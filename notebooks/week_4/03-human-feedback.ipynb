### Import Dependencies

```python
from langsmith import Client
```

### Attach the feedback to a specific trace in LangSmith

```python
client = Client()
```

```python
RUN_ID = "019f8f11-cf98-7202-b954-ff0a160dd54b" #TraceID from langsmith
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api",
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="this is a long form free text feedback left by me",
    feedback_source_type="api",
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment_with_score",
    score=1,
    value="this is a long form free text feedback left by me",
    feedback_source_type="api",
)
```
