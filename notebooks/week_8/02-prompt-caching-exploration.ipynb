```python
def count_tokens(text):
    encoding = tiktoken.get_encoding("o200k_base")
    tokens = encoding.encode(text)
    return len(tokens)

print(count_tokens(text))
```

### Instructor

```python
client = instructor.from_provider(
    "openai/gpt-5.4-mini"
)
```

### Completion Statistics

```python
def print_usage_and_cost_instructor(raw_response):

    input_price = 0.75/1000000
    output_price = 4.5/1000000
    cached_price = 0.075/1000000

    input_tokens = raw_response.usage.prompt_tokens
    output_tokens = raw_response.usage.completion_tokens
    cached_tokens = raw_response.usage.prompt_tokens_details.cached_tokens
    non_cached_tokens = input_tokens - cached_tokens

    print(f"Input tokens: {input_tokens}")
    print(f"Output tokens: {output_tokens}")
    print(f"Cached tokens: {cached_tokens}")
    print(f"Non-cached tokens: {non_cached_tokens}")

    print(f"Output cost: {output_tokens * output_price}")
    print(f"Cached cost: {cached_tokens * cached_price}")
    print(f"Non-cached cost: {non_cached_tokens * input_price}")
```

### Test 1
- when evaluating prompt caching, make sure the prompt is the same for both runs, otherwise the cache will be invalidated and the prompt will be seen by the model as new. A single space makes the prompt different, for example

```python
prompt_1 = f"""
Extract names from the following text :

{text}
"""
```

```python
prompt_2 = f"""
Extract places from the following text :

{text}
"""
```

```python
class ExtractedNames(BaseModel):
    names: list[str]

class ExtractedPlaces(BaseModel):
    places: list[str]
```

```python
_, raw_response_1 = client.create_with_completion(
    messages = [{"role": "system", "content": prompt_1}],
    response_model=ExtractedNames
)
```

```python
_, raw_response_2 = client.create_with_completion(
    messages = [{"role": "system", "content": prompt_2}],
    response_model=ExtractedPlaces
)
```

```python
print('Prompt 1 stats:')
print_usage_and_cost_instructor(raw_response_1)
print('---------------')
print('Prompt 2 stats:')
print_usage_and_cost_instructor(raw_response_2)
```

```python
_, raw_response_1 = client.create_with_completion(
    messages = [{"role": "system", "content": prompt_1}],
    response_model=ExtractedNames
)
```

```python
_, raw_response_2 = client.create_with_completion(
    messages = [{"role": "system", "content": prompt_2}],
    response_model=ExtractedPlaces
)
```

### LangChain prompt caching

#### 1. Exploring Langchain tool calling

```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, SystemMessage
```

```python
@tool
def dummy_tool(a: str, b: str) -> str:
    """
    Concatenate two strings.

    Args:
        a: First string
        b: Second string

    Returns:
        Concatenated string
    """
    return f"Hello {a} and {b}"
```

```python
llm = ChatOpenAI(model="gpt-5.4-mini")
llm_with_tools = llm.bind_tools([dummy_tool], tool_choice="required")
```

```python
response = llm_with_tools.invoke('Use dummy tool to say hello to John and Jane')

print(response)
```

```python
response.response_metadata['token_usage']
```

```python
def print_usage_and_cost_langchain(raw_response):

    input_price = 0.75/1000000
    output_price = 4.5/1000000
    cached_price = 0.075/1000000

    input_tokens = raw_response.response_metadata['token_usage']['prompt_tokens']
    output_tokens = raw_response.response_metadata['token_usage']['completion_tokens']
    cached_tokens = raw_response.response_metadata['token_usage']['prompt_tokens_details']['cached_tokens']
    non_cached_tokens = input_tokens - cached_tokens

    print(f"Input tokens: {input_tokens}")
    print(f"Output tokens: {output_tokens}")
    print(f"Cached tokens: {cached_tokens}")
    print(f"Non-cached tokens: {non_cached_tokens}")

    print(f"Output cost: {output_tokens * output_price}")
    print(f"Cached cost: {cached_tokens * cached_price}")
    print(f"Non-cached cost: {non_cached_tokens * input_price}")
```

```python
prompt_6 = f"""
<text>
{text}
</text>
"""
```

```python
llm = ChatOpenAI(model="gpt-5.4-mini")
llm_with_tools_1 = llm.bind_tools([ExtractedNames], tool_choice="required")
llm_with_tools_2 = llm.bind_tools([ExtractedPlaces], tool_choice="required")
```

```python
response = llm_with_tools_1.invoke(
    [
        SystemMessage(content=prompt_6),
        HumanMessage(content="Extract names from the text between the <text> tags.")
    ]
)
print_usage_and_cost_langchain(response)
```

```python
response = llm_with_tools_2.invoke(
    [
        SystemMessage(content=prompt_6),
        HumanMessage(content="Extract places from the text between the <text> tags.")
    ]
)
print_usage_and_cost_langchain(response)
```
